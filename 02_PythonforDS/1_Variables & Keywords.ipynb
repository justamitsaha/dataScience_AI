<h1 style = "color : Brown">Variables & Keywords</h1>

A Python variable is a reserved memory location to store values. In other words, a variable in a python program gives data to the computer for processing.

```python
#String
a='hello world'
type(a)
print(a)
```

```python
#Int
my_age = 31
print(my_age)
print(type(my_age))
```

```python
#Float
score = 89.996778
print(score)
print(type(score))
```

```python
#Boolean
a=True
print(type(a))
print(a)
```

```python
a = '10.5'
print(a)
print(type(a))
```

```python
a='True'
print(type(a))
print(a)
```

```python
# 5a = 10  Not allowed
a = 5*10;
a
```

```python
#Expression
a=4
b=5
c=a+b
print(c)
```

```python
#Function
def sum_fun(a,b):
    c=a+b
    return c

sum_fun(5,6)
```

```python
fn = 'Amit'
ln = 'Saha'
count =3
print(f'{fn}\t{ln}') # Name formatted with tab space
print("\n"*count, end="HI")
print()
print(fn, end="HI")
print()
print(count * fn)  # print number of times

```
