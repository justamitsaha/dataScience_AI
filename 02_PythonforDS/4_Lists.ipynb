<span style=color:orange><b>Indexing -->In Python List can be accesseed using +ve and -Ve indices both 
- String also supports this
</b></span>

```python
print(list2[0])
print(list2[-1])
print(list2[-2])
print(list2[-1][1])
print(list2[4][1])
print(list2[4][-1])
```

<span style=color:orange><b> Slicing</b></span>

```python
print(list2)
print(list2[2:])
print(list2[2:4])
print(list2[-3:-1])
print(list2[::])  # Entire list
print(list2[::2]) #Every 2nd element
print(list2[::-1])  #Reverse List
```

</b></span>

```python
list1 = ["Amit", "Ajey", "Alam", "Aveek"]
del list1[2] # will remove item at index 2
print(list1)
list1 = ["Amit", "Ajey", "Alam", "Aveek","Alan", "Alex"]
del list1[2:5] # will delete item at index 2, 3, 4
print(list1)
list1 = ["Amit", "Ajey", "Alam", "Aveek","Alan", "Alex"]
del list1[2:10] # doesn't throw index out of bound removes till end
print(list1)
```

<span style=color:orange><b>
pop()
</b></span>
- <span style=color:orange><b>
Deletes the object based on the index value 
</b></span>
- <span style=color:orange><b>
Default --> Removes the last element
</b></span>


```python
#pop() - By default, it deletes the last object from the list
#pop takes index as an input
list1 = [11,22,33,44,88,77]
list1.pop(2) # Will remove item at index 2
print(list1)
list1.pop() # Will remove last item
print(list1) 
```

<span style=color:orange><b>
Difference between sort & sorted
- sort changes the value of the item applied on
- sorted doesn't change the value we have to assign it to some variable to get sorted value
</b></span>

```python
A = ["Oranges", "Strawberry", "Mango"]
B = A.sort()    # doesn't return anything just updates values
print(A)
print(B)        # nothing got assigned
A = ["Oranges", "Strawberry", "Mango"]
B = sorted(A)   # returns values getting assigned to B
print(A)        #Values of A not chnaged
print(B)
```

<span style=color:orange><b>
Shallow Copy
</b></span>

```python
A = ["Oranges", "Strawberry", "Mango"]
B = A
print(A.pop())
print(B)        # B also modified
print(A)
A = ["Oranges", "Strawberry", "Mango"]
B = A[0:3]
print(A.pop())
print(B)        #B not modified
print(A)
```
