<span style=color:orange><b>Indexing -->In Python List can be accesseed using +ve and -Ve indices both </b></span>

```python
print(list2[0])
print(list2[-1])
print(list2[-2])
print(list2[-1][1])
print(list2[4][1])
print(list2[4][-1])
```

<span style=color:orange><b> Slicing</b></span>

```python
print(list2)
print(list2[2:])
print(list2[2:4])
print(list2[-3:-1])
print(list2[::])  # Entire list
print(list2[::2]) #Every 2nd element
print(list2[::-1])  #Reverse List
```
