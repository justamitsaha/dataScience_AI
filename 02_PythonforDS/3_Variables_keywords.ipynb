### <span style='color:orange'><b>What is a Variable?</b></span>
- A name that refers to a value in memory
- Python uses **dynamic typing** → type decided at runtime
- Variables are just **labels** pointing to objects

````python
x = 10
y = 'hello'
print(type(x), type(y))
````

### <span style='color:orange'><b>Variable Naming Rules</b></span>
- Must start with letter or underscore (`_`)
- Cannot start with digit
- Can contain letters, digits, `_`
- Case-sensitive (`age` != `Age`)
- Cannot use keywords

````python
print('valid:', 'my_var'.isidentifier())
print('invalid:', '2abc'.isidentifier())
````

### <span style='color:orange'><b>Multiple Assignment</b></span>
- Assign multiple variables in one line
- Swap without temp variable

````python
a, b = 10, 20
a, b = b, a
print(a, b)
````

### <span style='color:orange'><b>How Python Stores Variables</b></span>
- Variables store **references** to objects
- Small integers & strings → **interned** for efficiency

````python
a = 256
b = 256
print(a is b)  # Interned
x = 257
y = 257
print(x is y)  # Not always interned
````

### <span style='color:orange'><b>Mutable vs Immutable</b></span>
- Immutable: int, float, bool, str, tuple
- Mutable: list, dict, set, bytearray

````python
s = 'hello'
try:
    s[0] = 'H'
except Exception as e:
    print('Immutable:', e)
l = [1,2,3]
l[0] = 99
print(l)
````

### <span style='color:orange'><b>Variable Scope: LEGB</b></span>
- **L**ocal
- **E**nclosed
- **G**lobal
- **B**uilt-in

````python
x = 'global'

def outer():
    x = 'enclosed'
    def inner():
        x = 'local'
        print(x)
    inner()
outer()
````

### <span style='color:orange'><b>global & nonlocal Keywords</b></span>
- `global` → modify global variable
- `nonlocal` → modify enclosing scope variable

````python
x = 5

def test():
    global x
    x = 99

test()
print(x)
````

````python
def outer():
    a = 10
    def inner():
        nonlocal a
        a = 20
    inner()
    print(a)
outer()
````

### <span style='color:orange'><b>locals() vs globals()</b></span>
- `locals()` → variables in local scope
- `globals()` → module-level variables

````python
x = 10
def f():
    y = 20
    print('locals:', locals())

print('globals contains x?', 'x' in globals())
f()
````

<span style=color:orange><b>
Eval --> evaluates a string as a Python expression and returns the result
- It only works with expressions, not statements (e.g., you can’t use eval("for i in range(5): print(i)")).
- It can access variables in the current scope or in the dictionaries you pass.
- Security Warning: Using eval() on untrusted input is dangerous because it can execute arbitrary code.
</b></span>

````python
x = 10
result = eval("x + 5")
print(result)  # Output: 15
````

<span style=color:orange><b>
Eval has high security concern if mixed with User input
- Imagine you allow users to input a math expression:
- If the user enters: 
 __import__('os').system('rm -rf /') it can delete files
</b></span>

````python
import re

user_input = input("Enter an expression: ")

if user_input:
    # Allow digits, spaces, and basic operators (+, -, *, /)
    if re.fullmatch(r"[0-9+\-*/(). ]+", user_input):
        print(eval(user_input))
    else:
        print("Invalid expression! Only numbers and + - * / allowed.")
else:
    print("Please enter an expression.")
````

<span style=color:orange><b>
