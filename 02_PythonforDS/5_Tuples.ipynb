## Tuples

   - Tuples are ordered sequence of mixed data types
   - Tuples are written in comma separated elements within parenthesis

```python
t = ('hello', 5, 5.5)
print(type(t))
len(t)
```

```python
t1 = ('hello', 5, 5.5, [2,3,4])
var = t1[3]
type(var)
```

```python
var[1]
```

```python
t1[3][1]
```

```python
len(t1)
```

    Tuples can be defined in various ways

```python
t1 = 1,2,3,4,5
print(t1)
print(type(t1))
```

```python
#single value tuple
t1 = 1.2,
type(t1)
```

```python
#This is not a tuple

t4 = 1
t5 = (1)
```

```python
type(t5)
```

#### Indexing in tuple

```python
t = ("Mumbai", 10, "Python", 100)
```

```python
len(t)
```

```python
t[2:]
```

```python
str1 = "HELLO WORLD"
str1[6:]
```

```python
#gives the element at index location 0 i.e. 1st element

print(t[0])

#gives the element at index location 1

print(t[1])
```

#### Slicing

```python
#slicing first 2 elements from t

t = ("Mumbai", 10, "Python", 100)
print(t[0:2])
print(t[2:])
```

```python
t[-2:]
```

```python
#slicing last 2 elements from t

t[-2:]
```

```python
#find the number of elements in a tuple

len(t)
```

#### Concatenation

```python
tup1 = ("welcome", "to", "today's")
tup2 = ("class", "on", "python")
```

```python
tup3 = tup1 + tup2
```

```python
tup3
```

#### min, max, sum functions

```python
t = (2,4,6,8)
print(sum(t))
print(min(t))
print(max(t))
print(sorted(t))
```

#### Immutability

```python
t = ("Hello!!", "Welcome" , 5, 10, "world", "of", "Analytics!!")
# t[2] = "to" # Will give error as tuples are immutable
```

To update the 2nd and 3rd element we have to copy to new element

```python
new_tuple = t[0:2] + ("to", "the",) + t[-3:]
```
