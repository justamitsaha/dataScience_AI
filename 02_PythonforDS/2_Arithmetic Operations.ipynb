## <span style=color:cornflowerblue><b> Arithmetic Operations</b></span>

```python
a=10
b=2
print(a**b)
print(a/b)
print(a-b)
print(a//b)
print(a%b) # Modulus/Reminder
print(11%3)
```

```python
import math
a = 7/3
print(a)
print(math.floor(a))
print(math.ceil(a))
```


<span style=color:orange><b>
PEMDAS
1. Parentheses: ()
2. Exponentiation: **
3. Unary + and -: +x, -x
4. Multiplication / Division / Floor Division / Modulus: *, /, //, %
5. Addition and Subtraction: +, -
6. Bitwise Shifts: <<, >>
7. Bitwise AND: &
8. Bitwise XOR: ^
9. Bitwise OR: |
10. Comparisons: <, >, <=, >=, ==, !=
11. Boolean NOT: not
12. Boolean AND: and
13. Boolean OR: or
14. Assignment: =, +=, -=, etc.

These rules start with Parentheses, and then operations are performed on the exponents or powers. Next, we perform operations on multiplication or division from left to right. Finally, operations on addition or subtraction are performed from left to right.
</b></span>
