### 🟦 Python Data Types — Quick Reference

### 🔹 Strings
- Immutable sequences of Unicode characters
- Support indexing, slicing, concatenation, repetition
- Stored internally in Unicode
- **Immutability**: modifying characters requires creating new strings

```python
s = 'hello'
print(s[0])
# s[0] = 'H'  # ❌ Error: strings are immutable
s2 = 'H' + s[1:]
print(s2)
```

#### **String operations**
- `.upper()`, `.lower()`, `.strip()`, `.replace()`, `.split()`
- Slicing: `s[1:4]`, `s[::-1]` (reverse)
- Length: `len(s)`


```python
s= "Amit Saha"
print(s.upper())
print(s.lower())
print(s.replace("Amit", "AMIT"))
print(s.split(" "))
print(s.find("Saha"))
print(s.strip())  # removes leading/trailing whitespace
```

### 🔹 Integers
- Whole numbers
- Arbitrary precision (Python auto-expands storage)
- Supports binary, octal, hex literals

```python
a = 10
b = 0b1010  # binary
c = 0xFF     # hex
print(a, b, c)
```

### 🔹 Floats
- Decimal numbers (IEEE‑754 double precision)
- Approximation → rounding errors possible

```python
x = 0.1 + 0.2
print(x)  # 0.30000000000000004
```

### 🔹 Booleans
- `True`, `False`
- Returned by comparisons
- Subclass of `int` (`True = 1`, `False = 0`)

```python
print(True + True)
print(False + 5)
```

### 🔹 Typecasting
Convert across data types when compatible.

```python
print(type(int('12')))
print(type(float('3.14')))
print(type(str(999)))
string_var = "2.5"
#int(string_var) will give error as underlying is float
print(int(float("2.5")))
```

### 🔹 Bytes
- Immutable sequence of **0–255**
- Used for raw binary data, files, network streams
- Usually created by encoding a string (UTF‑8 default)

#### Examples

```python
b = b'hello'
print(b)
print(list(b))
```

While converting from bytes to String or vice-versa must mention encoding, Python won't know which encoding to use

```python
b = b'hello'
print(type(b))
print(b)
s = 'שלום'
b = bytes(s, encoding='utf-8')
#or
b = s.encode(encoding='utf-8')
#or
print(b)
s = str(b, encoding='utf-8')
print(s)
s = str(b, encoding='latin-1')
print(s) # gibberish
```

### 🔹 Bytearray
- Mutable version of bytes
- Useful for in-place modification like modifying image or network input 

```python
ba = bytearray(b'hello')
ba[0] = 72   # 'H'
print(ba)
```

### 🔹 Unicode & Encoding
- Python strings → Unicode
- To store/send → **encode** to bytes
- Common encodings: UTF‑8, UTF‑16, ASCII

UTF‑8 uses variable length:
- English letters → 1 byte
- European chars → 2 bytes
- Chinese/Japanese → 3 bytes
- Emojis → 4 bytes


```python
'A'.encode('utf-8')
'ñ'.encode('utf-8')
'中'.encode('utf-8')
'😀'.encode('utf-8')
```

### 🔹 Encoding vs Decoding
- `encode()` → string → bytes
- `decode()` → bytes → string
Must use **matching encoding**.

```python
b = 'café'.encode('utf-8')
print(b)
print(b.decode('utf-8'))
```
