```python
# Binomial Distribution Calculator

import math
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Function for binomial probability
def binomial_probability(n, x, p):
    """Calculate P(X=x) for binomial distribution"""
    nCx = math.comb(n, x)
    return nCx * (p**x) * ((1-p)**(n-x))

# User input
n = int(input("Enter number of trials (n): "))
p = float(input("Enter probability of success (p): "))
x = int(input("Enter number of successes (x): "))

# Probability of exactly x successes
manual_prob = binomial_probability(n, x, p)
scipy_prob = binom.pmf(x, n, p)

print(f"\nFor n={n}, p={p}, x={x}:")
print(f"Manual Calculation P(X={x}): {manual_prob}")
print(f"SciPy   Calculation P(X={x}): {scipy_prob}")

# Full distribution for 0...n
x_values = list(range(n+1))
probs = [binomial_probability(n, k, p) for k in x_values]

print("\nFull Probability Distribution:")
for k, prob in zip(x_values, probs):
    print(f"P(X={k}) = {prob:.5f}")

# Plot distribution
plt.figure(figsize=(8,5))
plt.bar(x_values, probs, color="skyblue", edgecolor="black")
plt.xlabel("Number of Successes (x)")
plt.ylabel("Probability P(X=x)")
plt.title(f"Binomial Distribution (n={n}, p={p})")
plt.show()
```
