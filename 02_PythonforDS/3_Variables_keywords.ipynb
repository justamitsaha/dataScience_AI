<span style=color:orange><b>
Passing Arguments
</b></span>

```python
import sys

# Simulate command-line arguments
sys.argv = ['notebook.ipynb', '7', '9']

# Validate and use arguments
if len(sys.argv) < 3:
    print("Please provide two numbers as arguments!")
else:
    try:
        x = int(sys.argv[1])
        y = int(sys.argv[2])
        print("Product:", x * y)
    except ValueError:
        print("Invalid input! Arguments must be integers.")
```

<span style=color:orange><b>
Eval --> evaluates a string as a Python expression and returns the result
- It only works with expressions, not statements (e.g., you can’t use eval("for i in range(5): print(i)")).
- It can access variables in the current scope or in the dictionaries you pass.
- Security Warning: Using eval() on untrusted input is dangerous because it can execute arbitrary code.
</b></span>

```python
x = 10
result = eval("x + 5")
print(result)  # Output: 15
```

<span style=color:orange><b>
Eval has high security concern if mixed with User input
- Imagine you allow users to input a math expression:
- If the user enters: 
 __import__('os').system('rm -rf /') it can delete files
</b></span>

```python
import re

user_input = input("Enter an expression: ")

if user_input:
    # Allow digits, spaces, and basic operators (+, -, *, /)
    if re.fullmatch(r"[0-9+\-*/(). ]+", user_input):
        print(eval(user_input))
    else:
        print("Invalid expression! Only numbers and + - * / allowed.")
else:
    print("Please enter an expression.")
```

<span style=color:orange><b>
exec() runs statements (good for loops, function definitions).
- This also has security risk if combigned with user input
</b></span>

```python

code_block = """
for i in range(3):
    print("exec prints:", i)
"""
exec(code_block)  # Output: 0, 1, 2
```

```python
fn = 'Amit'
ln = 'Saha'
count =3
print(f'{fn}\t{ln}')# Prints first name and last name separated by a tab space
print("\n"*count, end="HI") # Prints 3 newline characters, then ends with "HI" (instead of default newline)
print()
print(fn, end="HI")  # Prints first name and then "HI" without adding a newline at the end
print()
print(count * fn)  # print number of times
```
