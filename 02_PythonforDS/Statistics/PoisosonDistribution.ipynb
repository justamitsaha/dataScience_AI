```python
import math
import matplotlib.pyplot as plt

# Function to calculate Poisson probability
def poisson_probability(lmbda, x):
    return (lmbda**x * math.exp(-lmbda)) / math.factorial(x)

# Example: lambda = 10 (avg 10 calls per hour)
lmbda = 10  

# Calculate probabilities for 0 to 20 calls
x_values = list(range(0, 21))
probabilities = [poisson_probability(lmbda, x) for x in x_values]

# Print probabilities
for x, p in zip(x_values, probabilities):
    print(f"P(X={x}) = {p:.5f}")

# Plot distribution
plt.bar(x_values, probabilities, color="skyblue", edgecolor="black")
plt.title(f"Poisson Distribution (λ = {lmbda})")
plt.xlabel("Number of Calls (X)")
plt.ylabel("Probability P(X)")
plt.show()
# This script calculates and visualizes the Poisson distribution for a given lambda value.
```
